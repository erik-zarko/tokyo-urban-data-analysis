# Modules

```python
import pandas as pd
import folium
```

# Open Dataset

```python
df = pd.read_csv("../data/raw/tokyo_stations.csv")

print(df.head())
```

```python
df['prefecture'].unique()
```

```python
len(df['name'].unique())
```

# Creatie Interactive Map

```python
tokyo_map = folium.Map(
    location=[35.68, 139.76],
    zoom_start=11
)
```

```python
station_check = []
for _, row in df.iterrows():
    if row["name"] in station_check:
        pass
    else:
        station_check.append(row["name"])
        folium.CircleMarker(
            location=[row["y"], row["x"]],
            radius=3,
            popup=f"""
            <b>{row['name']}</b><br>
            {row['line']}
            """,
            fill=True
        ).add_to(tokyo_map)

print(len(station_check))
```

```python
tokyo_map.save("../imgs/tokyo_stations_map.html")
```
